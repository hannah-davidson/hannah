# 2. Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    return A * np.sin(w * x)
```

```python
x = np.linspace(0, 2*np.pi, 1000)
```

```python
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
```

```python
fig, axes = plt.subplots(1, 3, figsize=(25, 5))

for A, w in zip(amplitude, frequency):
    axes[0].plot(x, make_sine_wave(x, A=A, w=1), label=f"A={A}")

for A, w in zip(amplitude, frequency):
    axes[1].plot(x, make_sine_wave(x, A=1, w=w), label=f"w={w}")

for A, w in zip(amplitude, frequency):
    axes[2].plot(x, make_sine_wave(x, A=A, w=w), label=f"A={A}, w={w}")

axes[0].set_title("Sine Function, Varying Amplitude")
axes[0].set_xlabel("x")
axes[0].set_ylabel("f(x)")
axes[0].legend()

axes[1].set_title("Sine Function, Varying Frequency")
axes[1].set_xlabel("x")
axes[1].set_ylabel("f(x)")
axes[1].legend()

axes[2].set_title("Sine Function, Varying Amplitude and Frequency")
axes[2].set_xlabel("x")
axes[2].set_ylabel("f(x)")
axes[2].legend()

plt.show()

# When everything is in one cell, Python runs it top to bottom without interruption:
# All in one cell — labels are set BEFORE legend is called
# axes[0].plot(x, make_sine_wave(x, A=A, w=1), label=f"A={A}")  # label set
# axes[0].legend()  # finds the label ✓
```
