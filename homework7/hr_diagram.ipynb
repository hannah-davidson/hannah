```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("/Users/hannahdavidson/python_decal_sp26/course_assignments/homework7/star_data.txt",skiprows=1)
```

```python
print(np.shape(data))
# There are 2642 rows and 2 columns.
# first column: magnitude of star in units of magnitude
# second column: temperature of star in units of Kelvin
# each row in this dataset is a different star
```

```python
magnitude = data[:,0]
temperature = data[:,1]
```

```python
print(f"The minimum magnitude is {np.min(magnitude)}, corresponding to the {np.argmin(magnitude) + 1}th star in the list.")
print(f"The maximum magnitude is {np.max(magnitude)}, corresponding to the {np.argmax(magnitude) + 1}th star in the list.")
```

```python
print(f"The minimum temperature is {np.min(temperature)} K, corresponding to the {np.argmin(temperature) + 1}th star in the list.")
print(f"The maximum temperature is {np.max(temperature)} K, corresponding to the {np.argmax(temperature) + 1}th star in the list.")
```

Since larger numbers = dimmer stars, the brightest star is the one with the minimum magnitude, and the dimmest star is the one with the minimum magnitude.

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
```

```python
plt.figure(figsize = (8,6))

plt.scatter(log_temperature, log_magnitude)
plt.scatter(np.log(temperature[1988]), np.log(np.min(magnitude)), color='k', marker='*',s=300, label = 'brighest star') # brightest star
plt.scatter(np.log(temperature[264]), np.log(np.max(magnitude)), color='purple', marker='*', s=300, label = 'dimmest star') # dimmest star
plt.scatter(np.log(np.max(temperature)), np.log(magnitude[2613]), color='g', marker='*', s=300, label = 'hottest star') # hottest star
plt.scatter(np.log(np.min(temperature)), np.log(magnitude[1923]), color='orange', marker='*',s=300, label = 'coldest star') # coldest star
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.title("Hertzsprung-Russell Diagram")

plt.legend()
plt.show()
```
