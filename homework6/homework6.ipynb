## 3.2 2D NumPy Arrays

```python
matrix = [
    [5, 12, 17, 9],
    [8, 21, 14, 3],
    [6, 13, 19, 11]
]

Matrix = np.array(matrix)
```

```python
print(f"Matrix has {Matrix.shape[0]} rows and {Matrix.shape[1]} columns.")
```

```python
print(Matrix[:, 2])
```

```python
print(Matrix[1, :])
```

```python
print(Matrix*2)
```

```python
for col in Matrix.T:
    column_mean = np.mean(col)
    print(column_mean)
```

```python
mean_array = np.mean(Matrix)
mean_array
```

```python
for row in Matrix:
    max_val = np.max(row)
    print(max_val)
```

```python
for column in Matrix.T: # have to do transpose!
    min_val = np.min(column)
    print(min_val)
```

```python
print(Matrix.flatten()) # Makes it a 1D array.
```

```python
argmin = np.argmin(Matrix.flatten())
```

```python
argmax = np.argmax(Matrix.flatten())
```

```python
max_pos = np.unravel_index(argmax, Matrix.shape)
min_pos = np.unravel_index(argmin, Matrix.shape)
```

```python
filtered = Matrix[Matrix > 15]
print("Values greater than 15:", filtered)
```

```python
sorted_matrix = np.sort(Matrix, axis=1)
print("Sorted matrix:\n", sorted_matrix)
```

# 4 Prime Star Systems

```python
# this isn't the name of the variable it says to use in the homework but for the 
# sake of preventing redunancy...
star_arr = np.array([
    [2,3,5],
    [14,6,8],
    [11,13,17],
    [7,10,13]
])
```

```python
def is_prime(n):
    if n < 2:
        return False

    sieve = [True] * (n + 1)
    sieve[0] = sieve[1] = False

    for i in range(2, int(n**0.5) + 1):
        if sieve[i]:
            for j in range(i * i, n + 1, i):
                sieve[j] = False

    return sieve[n]
```

```python
def print_prime_rows(array):
    for row in star_arr:
        for i in row:
            if is_prime(i) == True:
                print(row)
                break
```

```python
print_prime_rows(star_arr)
```

# 5 Let's Play Checkers!

```python
zeroes = np.zeros(shape=(8,8))
zeroes
```

```python
zeroes[0::2, ::2] = 1         # odd rows, every other column
print(zeroes)
```

```python
zeroes[1::2, 1::2] = 1
print(zeroes)
```

```python
reverse_check = np.abs(zeroes - 1)
print(reverse_check)
```

# 7 Second-Brightest Star

```python
stars = np.random.randint(500, 2000, (5,5))
stars
```

```python
def second_brightest(array):
    for row in array:
        second_brightest = sorted(row, reverse=True)
        print(second_brightest[1])

second_brightest(stars)
```
